# 01 — Data Exploration

Sanity checks on captured tick data.

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))

import pandas as pd
from backend.storage.tick_store import TickStore
from backend.config import settings

store = TickStore(settings.tick_store_dir)
today = pd.Timestamp.today().strftime('%Y-%m-%d')
df = store.read_day('RELIANCE', today)
df.head()
```

```python
df['spread'] = df['ask_px_1'] - df['bid_px_1']
df.set_index('ts')['spread'].plot(figsize=(12, 3), title='Quoted spread over the session')
```
